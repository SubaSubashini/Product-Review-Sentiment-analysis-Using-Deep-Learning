```python
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Embedding,LSTM
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
train_data, test_data=train_test_split(data,test_size=0.2,random_state=42)
```

```python
train_data.shape
```

```python
test_data.shape
```

```python
tokenizer=Tokenizer(num_words=5000)
# Fill missing values in 'Review' column with empty strings before tokenizing
train_data["Review"] = train_data["Review"].fillna("")
tokenizer.fit_on_texts(train_data["Review"])
```

```python
test_data["Review"] = test_data["Review"].fillna("") # Fill missing values in 'Review' column with empty strings
x_train=pad_sequences(tokenizer.texts_to_sequences(train_data["Review"]),maxlen=200)
x_test=pad_sequences(tokenizer.texts_to_sequences(test_data["Review"]),maxlen=200)
```

```python
x_train
#its give me a array formets like..
#I LOVE ANALYSIS CONVERT TO
#1  2     3
```

```python
x_test
```

```python
y_train=train_data["sentiment"].replace({"positive":1,"negative":0})
y_test=test_data["sentiment"].replace({"positive":1,"negative":0})
```

```python
y_train
```

```python
#LSTM MODEL BUILDING
```

```python
model=Sequential()
model.add(Embedding(input_dim=5000,output_dim=128,input_length=200))
model.add(LSTM(128,dropout=0.2,recurrent_dropout=0.2))
model.add(Dense(1,activation="sigmoid"))
```

model.summary

```python
model.compile(optimizer="adam",loss="binary_crossentropy",metrics=["accuracy"])
```

```python
model.fit(x_train,y_train,epochs=5,batch_size=64,validation_split=0.2)
```

```python
print("Data type of y_train:", y_train.dtype)
print("First 5 values of y_train:", y_train.head())
print("\nData type of y_test:", y_test.dtype)
print("First 5 values of y_test:", y_test.head())
```

```python
model.save("model.h5")
```

```python
import joblib
joblib.dump(tokenizer,"tokenizer.pk1")
```

```python
loss,accuracy=model.evaluate(x_test,y_test)
```

```python
import gradio as gr
import random

# -----------------------------
# 🧠 Dummy sentiment analysis function
# -----------------------------
def sentiment_analysis(review):
    review = review.lower()
    if any(word in review for word in ["good", "excellent", "amazing", "love", "best"]):
        return "😊 Positive Review"
    elif any(word in review for word in ["bad", "worst", "terrible", "hate", "poor"]):
        return "😡 Negative Review"
    else:
        return "😐 Neutral Review"

# -----------------------------
# 🎨 Custom CSS for styling
# -----------------------------
custom_css = """
body {
    background: linear-gradient(135deg, #0f2027, #203a43, #2c5364);
    color: white;
}

.gradio-container {
    background-color: transparent !important;
}

#submit-btn {
    background-color: #ff9800 !important;
    color: white !important;
    font-weight: bold !important;
    border-radius: 12px !important;
    transition: 0.3s ease-in-out !important;
}
#submit-btn:hover {
    background-color: #ffb74d !important;
    transform: scale(1.05);
}

#clear-btn {
    background-color: #607d8b !important;
    color: white !important;
    font-weight: bold !important;
    border-radius: 12px !important;
}
#clear-btn:hover {
    background-color: #90a4ae !important;
}

h1 {
    text-align: center;
    font-size: 28px;
    color: #ffcc80;
}
"""

# -----------------------------
# 🧩 Build the interface
# -----------------------------
with gr.Blocks(css=custom_css, theme=gr.themes.Soft()) as app:
    gr.HTML("<h1>🛍️ PRODUCT REVIEW SENTIMENT ANALYSIS APPLICATION</h1>")

    with gr.Row():
        review = gr.Textbox(
            label="Enter your Product Review ✍️",
            placeholder="Type your review here...",
            lines=3
        )

    with gr.Row():
        submit = gr.Button("Submit", elem_id="submit-btn")
        clear = gr.Button("Clear", elem_id="clear-btn")

    output = gr.Textbox(label="Predicted Sentiment 💬")

    submit.click(sentiment_analysis, inputs=review, outputs=output)
    clear.click(lambda: "", None, review)

# -----------------------------
# 🚀 Launch app
# -----------------------------
app.launch(share=True)

```
